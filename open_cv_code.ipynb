# Getting Started With OpenCV

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Reading and Displaying an Image

```python
img = cv2.imread("sample.jpg")
if img is None:
    print("Image could not be loaded. Check Filepath")
else:
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(6, 4))
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.axis("off")
    plt.show()
```

## 2. Inspecting image properties

```python
if img is not None:
    print("Shape:", img.shape)
    print("Height:", img.shape[0])
    print("Width:", img.shape[1])
    print("Number of channels:", img.shape[2])
    print("Data type:", img.dtype)
    print("Pixel at row 50, column 100:", img[50, 100])
```

## 3. Convert Image to Grayscale

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print("Original color shape:", img.shape)
print("Grayscale shape:     ", gray.shape)

plt.figure(figsize=(6, 4))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

## 4. Saving an Image

```python
if img is not None:
    success = cv2.imwrite("copy.png", img)
    print("Saved successfully:", success)
```

## 5. Resizing Image

```python
resized = cv2.resize(img, (300, 200), interpolation=cv2.INTER_AREA)

print("Original shape:", img.shape)
print("Resized shape: ", resized.shape) 

resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(5, 3))
plt.imshow(resized_rgb)
plt.title("Resized (300w x 200h)")
plt.axis("off")
plt.show()
```

## 6. Cropping an Image

```python
# Select rows 50 to 250 and columns 100 to 350
cropped = img[50:250, 100:350]

# Display inline with Matplotlib (BGR -> RGB)
cropped_rgb = cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(5, 3))
plt.imshow(cropped_rgb)
plt.title("Cropped Region")
plt.axis("off")
plt.show()
```

## 7. Blurring with Gaussian blur

```python
blurred = cv2.GaussianBlur(img, (7, 7), 0)
plt.imshow(blurred)
```

## 8. Median blur

```python
median = cv2.medianBlur(img, 5)
plt.imshow(median)
```

## 9. Binary thresholding

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

threshold_value, binary = cv2.threshold(
    gray, 127, 255, cv2.THRESH_BINARY)

plt.imshow(binary)
```

## 10. Adaptive thresholding

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

adaptive = cv2.adaptiveThreshold(
    gray, 255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11, 2
)

plt.imshow(adaptive)
```

## 11. Edge detection with Canny

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, 100, 200)

plt.imshow(edges)
```

## 12. Drawing lines, rectangles, and circles

```python
cv2.line(img, (20, 20), (250, 20), (0, 255, 0), 3)
cv2.rectangle(img, (50, 50), (250, 200), (255, 0, 0), 2)
cv2.circle(img, (150, 125), 40, (0, 0, 255), -1)

plt.imshow(img)
plt.title("Drawing")
```

## 13. Adding text to an image

```python
cv2.putText(
    img, "OpenCV Basics",
    (30, 50),
    cv2.FONT_HERSHEY_SIMPLEX,
    1, (0, 255, 0), 2
)

plt.imshow(img)
plt.title("Text Annotation")
```

## 14. Morphological Operations: Erosion and Dilation

```python
img = cv2.imread("sample.jpg", cv2.IMREAD_GRAYSCALE)
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

kernel = np.ones((3, 3), np.uint8)

eroded = cv2.erode(binary, kernel, iterations=1)
dilated = cv2.dilate(binary, kernel, iterations=1)

plt.imshow(binary)
plt.title("Binary")

plt.imshow(dilated)
plt.title("Dilation")

plt.imshow(eroded)
plt.title("Dilation")
```
